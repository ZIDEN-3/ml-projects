```python
import pandas as pd
import numpy as np

train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
X_Titanic = train_df.drop('Survived', axis=1)
y_Titanic = train_df['Survived']
X_Titanic.drop('Cabin', axis=1)
print(X_Titanic.shape)
print(y_Titanic.shape)
X_Titanic
```

```python
from sklearn.model_selection import train_test_split
Xtrain, Xtest, ytrain, ytest = train_test_split(X_Titanic, y_Titanic, train_size=0.5)
```

```python
print(len(Xtrain))
```

```python
print(Xtrain.isnull().sum().sum())
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(Xtrain, ytrain)
y_model = model.predict(Xtest)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(ytest, y_model)
```
