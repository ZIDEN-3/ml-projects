```python
import pandas as pd
import numpy as np

train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
answers_df = pd.read_csv('gender_submission.csv')
X_Titanic_train = train_df.drop(columns=['Survived', 'Name', 'Ticket', 'Cabin', 'Embarked'])
X_Titanic_train['Age'] = X_Titanic_train['Age'].fillna(X_Titanic_train['Age'].median())
#X_Titanic_train['Embarked'] = X_Titanic_train['Embarked'].fillna('Unknown')

y_Titanic_train = train_df['Survived']

X_Titanic_test = test_df.drop(columns=['Name', 'Ticket', 'Cabin', 'Embarked'])
X_Titanic_test['Age'] = X_Titanic_test['Age'].fillna(X_Titanic_test['Age'].median())
X_Titanic_test['Fare'] = X_Titanic_test['Fare'].fillna(X_Titanic_test['Fare'].median())
y_Titanic_test = answers_df.drop('PassengerId', axis=1)

print(X_Titanic_train.shape)
print(y_Titanic_train.shape)
print(X_Titanic_test.shape)
print(y_Titanic_test.shape)
X_Titanic_test
```

```python
print(len(X_Titanic_train) == len(X_Titanic_train))
```

```python
print(type(X_Titanic_test['Sex']))
print(X_Titanic_test['Sex'].dtype)
print(X_Titanic_test['Sex'].unique())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
X_Titanic_train['Sex'] = le.fit_transform(X_Titanic_train['Sex'])
X_Titanic_test['Sex'] = le.transform(X_Titanic_test['Sex'])
```

```python
print(X_Titanic_train.isnull().sum())
print(X_Titanic_test.isnull().sum())
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(X_Titanic_train, y_Titanic_train)
y_model = model.predict(X_Titanic_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_Titanic_test, y_model)
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(
    n_estimators=500, 
    min_samples_split=15,
    max_depth=10,
    max_features='sqrt')

model.fit(X_Titanic_train, y_Titanic_train)
y_model = model.predict(X_Titanic_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_Titanic_test, y_model)
```
